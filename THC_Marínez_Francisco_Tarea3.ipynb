```python
#Rectas
print("Ingresa pendiente (m) y ordenada al origen (b) separadas por espacio.")
b1, m1 = map(float, input("pendiente, ordenada").split())
b2, m2 = map(float, input("pendiente, ordenada").split())

if m1 == m2:
    if b1 == b2:
        print("son coincidentes")
    else:
        print("son paralelas")
else:
    x = (b2 - b1) / (m1 - m2)
    y = m1 * x + b1
    print("se intersectan en el punto:")
    print(f"({x}, {y})")
```

```python
#Perímetro de una Elipse. Aproximación de Ramanujan
import math
a = 8
b = 5
print(f"Para saber el perimetro de una elipse la aproximacion de ramanujan dice que...")
print(f"si a = {a} y b = {b} entonces el perimetro es:")

if a == b:
    print(f"{(math.pi * a * 2)}")
else:
  r = math.sqrt(((3 * a) + b) * ((3 * b) + a))
  l = (3 * (a + b))
  resultado = math.pi * (l - r)
  print(f"{resultado:.4f}")

print("pero, si:")
c = int(input("Ingresa el valor de a: "))
d = int(input("Ingresa el valor de b: "))

print(f"Entonces para cuando a = {c} y b = {d} entonces el perimetro es:")

if c == d:
    print(f"{(math.pi * c * 2):.4f}")
else:
  r2 = math.sqrt(((3 * c) + d) * ((3 * d) + c))
  l2 = (3 * (c + d))
  result = math.pi * (l2 - r2)
  print(f"{result:.4f}")
```

```python
#Desigualdad triangular y fórmula de Herón
a = int(input("Ingresa el valor de a: "))
b = int(input("Ingresa el valor de b: "))
c = int(input("Ingresa el valor de c: "))

if a + b > c and a + c > b and b + c > a:
    print("Si es un triangulo")

    if a == b == c:
        print("Es equilatero")

    elif a == b or a == c or b == c:
        print("Es isoceles")

    else:
        print("Es escaleno")

    s = (a + b + c) / 2
    area = (s * (s - a) * (s - b) * (s - c)) ** 0.5
    print(f"El area del triangulo es: {area:.4f}")
    print(f"El perimetro del triangulo es: {a + b + c}")
else:
    print("No es triangulo, vuelve a intentarlo")
```

```python
#Convertidor de Unidades del sistema inglés al S.I
print("¿Que quieres convertir? escribe 1, 2 o 3 dependiendo de lo que quieras.")
x = int(input("Longitud: 1, Masa: 2, Tiempo: 3"))
if x == 1 or x == 2 or x == 3:
  if x == 1:
    L = float(input("Dime tu longitud en millas."))
    if L >= 0:
      print(f"{L} millas son {L*1609.34} metros")
    else:
      print("No se puede convertir un numero negativo")
  elif x == 2:
    M = float(input("Dime tu masa en libras."))
    if M >= 0:
      print(f"{M} libras son {M*0.453592} kilos")
    else:
      print("No se puede convertir un numero negativo")
  else:
    print("Dame tu tiempo primero como los dias enteros y despues las horas que sobran")
    Td = int(input("Dime tu tiempo en dias."))
    Th = float(input("Dime tu tiempo en horas."))
    if Td >= 0 and Th >= 0:
      H = Td * 24 + Th
      print(f"{H} horas son {H*3600} segundos")
    else:
      print("No se puede convertir un numero negativo")
else:
  print(f"{x} no es una opción, digite una de las 3 opciones.")
```

```python
#Volumen y Área de la Superficie de una Pirámide Regular Recta
import math
h = float(input("Ingresa la altura de la piramide: "))
n = float(input("Ingresa los lados de la base: "))
s = float(input("Ingresa la longitud de un lado: "))

if h > 0 and n >= 3 and s > 0:
 a = 1/2 * s * math.atan(math.pi/n)
 A = 1/2 * n * s * a
 l = math.sqrt((h**2) + (a**2))
 V = 1/3 * A * h
 S = A + 1/2 * n * s * l
 print(f"El volumen de la piramide es: {V:.3f}")
 print(f"El area de la superficie es: {S:.3f}")
else:
  print("Error: Piramide no valida")
```
