```python
import math
#semiejes
a = 6378137 #mayor en metros
c = 6356752.314245 #menor en metros
r = 6371000 #radio en km a m

w = 1-(c**2/a**2)

e= math.sqrt(w)

Sobl = 2 * math.pi * (a**2) * (1 + ((1 - (w)) / e) * math.atanh(e))

Sesfera = 4 * math.pi * r * r

dif = ((Sesfera - Sobl) / Sesfera) * 100

print(dif)
```

```python
#La secuencia del cortador de pizza perezoso ( The Lazy Caterer's sequence)
n = int(input("Elige cuántos cortes hacer "))
fat = 1
print(f"Sin cortes entonces son {fat} rebanadas")
for i in range(1, n + 1):
    fac = fat + i
    print(f"Para {i} cortes entonces son {fac} rebanadas")
    fat = fac
```
